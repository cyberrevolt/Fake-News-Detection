```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../dataset/cleaned_news_dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df.columns.tolist()
```

```python
print(df.isnull().sum())
```

```python
X = df["clean_content"]
y = df["label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Total samples:", len(X))
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("Training labels:")
print(y_train.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())
```

```python
tfidf = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)
```

```python
X_train_tfidf = tfidf.fit_transform(X_train)
```

```python
print("Training TF-IDF shape:", X_train_tfidf.shape)
```

```python
X_test_tfidf = tfidf.transform(X_test)
```

```python
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
feature_names = tfidf.get_feature_names_out()

print("Number of features:", len(feature_names))
```

```python
print(feature_names[:50])
```

```python
sample_vector = X_train_tfidf[0]

print("Shape:", sample_vector.shape)
```

```python
print(sample_vector)
```

```python
import numpy as np

row = X_train_tfidf[0].toarray().flatten()

top_indices = row.argsort()[-20:][::-1]

for index in top_indices:
    if row[index] > 0:
        print(
            feature_names[index],
            "→",
            round(row[index], 4)
        )
```

```python
print("X_train_tfidf:", X_train_tfidf.shape)
print("X_test_tfidf:", X_test_tfidf.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
import joblib

joblib.dump(
    tfidf,
    "../model/tfidf_vectorizer.pkl"
)

print("TF-IDF vectorizer saved successfully!")
```

```python
train_data = pd.DataFrame({
    "text": X_train,
    "label": y_train
})

test_data = pd.DataFrame({
    "text": X_test,
    "label": y_test
})
```

```python
train_data.to_csv(
    "../dataset/train_data.csv",
    index=False
)

test_data.to_csv(
    "../dataset/test_data.csv",
    index=False
)

print("Training and testing datasets saved!")
```
